# Re-encode the globe movie
Joins the frames rendered by `sigma2_globe_movie.ipynb` into a faster movie, without rendering anything again.

```python
import subprocess

from IPython.display import Video
```

## Settings

```python
FRAME_DIR = "../data/globe_frames"           # frames from sigma2_globe_movie.ipynb (one per day)
BASE_DAYS_PER_ANIMATION_SECOND = 7.5         # speed used in sigma2_globe_movie.ipynb
SPEEDUP = 1.5
INTERPOLATE = False                          # blend in-between frames up to 30 fps (ffmpeg minterpolate)

MOVIE = f"../plots/ssigma2_antarctica_{SPEEDUP}x.mp4"
```

## Movie

```python
days_per_animation_second = BASE_DAYS_PER_ANIMATION_SECOND * SPEEDUP   # one frame per day, so also the frame rate
interpolation = ["-vf", "minterpolate=fps=30:mi_mode=mci"] if INTERPOLATE else []

subprocess.run([
    "ffmpeg", "-y", "-loglevel", "error",
    "-framerate", str(days_per_animation_second),
    "-i", f"{FRAME_DIR}/frame_%04d.png",
    *interpolation,
    "-c:v", "libx264", "-crf", "18", "-pix_fmt", "yuv420p",   # H.264, plays everywhere
    MOVIE,
], check=True)
print("wrote", MOVIE)

Video(MOVIE, width=960)
```
